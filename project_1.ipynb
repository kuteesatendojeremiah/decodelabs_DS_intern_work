```python
import os
import pandas as pd

REPO_DIR = "/content/DS_intern_work"
DATA_FILE = os.path.join(REPO_DIR, "Dataset for Data Analytics.xlsx")

# Clone the repo if it is not already present
if not os.path.exists(DATA_FILE):
    !git clone https://github.com/kuteesatendojeremiah/DS_intern_work.git

df = pd.read_excel(DATA_FILE)

print(f"Loaded {len(df):,} rows and {len(df.columns)} columns")
df.head()
```

```python
missing_val = df.isnull().mean()
print(missing_val)
cols_to_clean = missing_val[missing_val < 0.05].index
df_cleaned = df.dropna(subset = cols_to_clean)
print(df_cleaned)
```

Working on the CouponCode column since it has missing values


```python
mode_coupon_code = df['CouponCode'].mode()[0]
print(f"The mode of 'CouponCode' is: {mode_coupon_code}")
df['CouponCode'] = df['CouponCode'].fillna(mode_coupon_code)
df['CouponCode'].isna().sum()
```

```python
#isolating and neutralizing outliners
import numpy as np
Q1 = df['UnitPrice'].quantile(0.25)
Q3 = df['UnitPrice'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df['UnitPrice'] = np.clip(df['UnitPrice'], lower_bound, upper_bound)
df['UnitPrice'].min(), df['UnitPrice'].max()
((df['UnitPrice'] < lower_bound) | (df['UnitPrice'] > upper_bound)).sum()
```

## Phase 2: Vectorized Computation Engine - Feature Engineering

Now we'll extract new features from existing columns. Date-based features are often very insightful.

```python
df['Date'] = pd.to_datetime(df['Date'])

df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
df['DayOfWeek'] = df['Date'].dt.dayofweek # Monday=0, Sunday=6
df['WeekOfYear'] = df['Date'].dt.isocalendar().week.astype(int)
df['DayOfYear'] = df['Date'].dt.dayofyear

print("DataFrame with new date features:")
display(df.head())
```

### Aggregation-Based Features

creating features by aggregating data based on `Product` and `CustomerID`. These features can reveal patterns in customer behavior and product popularity.

```python
# Feature: Total Quantity Sold per Product
total_quantity_per_product = df.groupby('Product')['Quantity'].sum().rename('TotalQuantitySoldPerProduct').reset_index()
df = df.merge(total_quantity_per_product, on='Product', how='left')

# Feature: Average Unit Price per Product
avg_unit_price_per_product = df.groupby('Product')['UnitPrice'].mean().rename('AverageUnitPricePerProduct').reset_index()
df = df.merge(avg_unit_price_per_product, on='Product', how='left')

# Feature: Number of Orders per Customer
orders_per_customer = df.groupby('CustomerID')['OrderID'].nunique().rename('OrdersPerCustomer').reset_index()
df = df.merge(orders_per_customer, on='CustomerID', how='left')

# Feature: Average Items in Cart per Customer
avg_items_in_cart_per_customer = df.groupby('CustomerID')['ItemsInCart'].mean().rename('AverageItemsInCartPerCustomer').reset_index()
df = df.merge(avg_items_in_cart_per_customer, on='CustomerID', how='left')

print("DataFrame with new aggregation features:")
display(df.head())
```

## Phase 3: Structural Contracts and Scaling

This phase focuses on preparing the data for machine learning models, ensuring it adheres to structural contracts and is scaled appropriately. We'll start by identifying categorical and numerical features.

```python
categorical_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()

print(f"Categorical Columns: {categorical_cols}")
print(f"Numerical Columns: {numerical_cols}")
```

### Handling High Cardinality Categorical Features (Identifiers)

Columns like `OrderID` and `TrackingNumber` are unique identifiers for each transaction or tracking. They typically don't provide predictive value in their raw form for most machine learning tasks and can create issues with one-hot encoding due to their high cardinality. `CustomerID` was used to create aggregated features, but the original `CustomerID` column itself might not be directly useful as a categorical feature if it has too many unique values and we already have customer-level aggregates. For now, let's remove `OrderID` and `TrackingNumber`.

```python
cols_to_drop_identifiers = ['OrderID', 'TrackingNumber']
df = df.drop(columns=cols_to_drop_identifiers)

# Re-evaluate categorical and numerical columns after dropping identifiers
categorical_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()

print(f"Updated Categorical Columns: {categorical_cols}")
print(f"Updated Numerical Columns: {numerical_cols}")
print("DataFrame head after dropping identifier columns:")
display(df.head())
```

```python
cols_to_drop_high_cardinality = ['CustomerID', 'ShippingAddress']
df = df.drop(columns=cols_to_drop_high_cardinality)

# Re-evaluate categorical and numerical columns again
categorical_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()

print(f"Final Categorical Columns for Encoding: {categorical_cols}")
print(f"Final Numerical Columns for Scaling: {numerical_cols}")
print("DataFrame head after dropping high-cardinality columns:")
display(df.head())
```

### Encoding Categorical Features

Now that we've handled high-cardinality identifiers, we can encode the remaining categorical features using one-hot encoding. This converts categorical variables into a numerical format suitable for machine learning algorithms without implying any ordinal relationship.

```python
df_encoded = pd.get_dummies(df, columns=categorical_cols, drop_first=True)

print("DataFrame head after one-hot encoding:")
display(df_encoded.head())
print(f"New shape of DataFrame: {df_encoded.shape}")
```

### Scaling Numerical Features

Numerical features often have different scales and ranges, which can negatively impact the performance of many machine learning algorithms. We will use `StandardScaler` to transform these features so they have a mean of 0 and a standard deviation of 1.

```python
from sklearn.preprocessing import StandardScaler

# Identify numerical columns in the encoded DataFrame (excluding 'Date' if it's still present)
# 'Date' column is still present in df_encoded and should not be scaled, so we will drop it first
df_scaled = df_encoded.drop(columns=['Date']).copy()

# Re-identify numerical columns after dropping 'Date'
numerical_cols_final = df_scaled.select_dtypes(include=['int64', 'float64']).columns.tolist()

scaler = StandardScaler()
df_scaled[numerical_cols_final] = scaler.fit_transform(df_scaled[numerical_cols_final])

print("DataFrame head after scaling numerical features:")
display(df_scaled.head())
print(f"Final shape of prepared DataFrame: {df_scaled.shape}")
```

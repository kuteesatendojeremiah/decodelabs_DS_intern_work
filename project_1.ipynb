```python
import os
import pandas as pd

REPO_DIR = "/content/DS_intern_work"
DATA_FILE = os.path.join(REPO_DIR, "Dataset for Data Analytics.xlsx")

# Clone the repo if it is not already present
if not os.path.exists(DATA_FILE):
    !git clone https://github.com/kuteesatendojeremiah/DS_intern_work.git

df = pd.read_excel(DATA_FILE)

print(f"Loaded {len(df):,} rows and {len(df.columns)} columns")
df.head()
```

```python
# Quick overview
print("Shape:", df.shape)
print("\nColumns:", list(df.columns))
print("\nMissing values:\n", df.isnull().sum())
df.describe(include="all")
```

```python
missing_val = df.isnull().mean()
print(missing_val)
cols_to_clean = missing_val[missing_val < 0.05].index
df_cleaned = df.dropna(subset = cols_to_clean)
print(df_cleaned)
```

Working on the CouponCode column since it has missing values


```python
mode_coupon_code = df['CouponCode'].mode()[0]
print(f"The mode of 'CouponCode' is: {mode_coupon_code}")
df['CouponCode'] = df['CouponCode'].fillna(mode_coupon_code)
df['CouponCode'].isna().sum()
```
